# **Pyplot API**

# **1.Univariate - Numeric**

In Matplotlib, a univariate plot means you’re visualizing the distribution or pattern of one variable only.

The idea is: you take a single column of data and display how its values are spread out, counted, or summarized.

```python
data = {
    "Salary" : [25000, 30000, 37000, 28000, 39000, 48000, 55000, 52000, 35000, 26000, 27000, 31000, 33000, 29000, 40000, 36000, 34000, 45000, 47000, 50000]
}
df = pd.DataFrame(data)
```

# **Line Plot**

A line plot is one of the most basic and important chart types — it connects data points with straight lines to show trends or changes over a continuous range, usually time or some other ordered sequence.

## **Key idea**

**X-axis:** the independent variable (often time, but can be anything ordered — distance, age, index)

**Y-axis:** the dependent variable (the single variable we’re measuring)

**Points:** actual recorded measurements

**Lines:** show progression or trend between points

## **Why use a line plot**

To see trends over time

To show patterns like growth, decline, or cycles

To compare multiple variables (by plotting multiple lines)

```python
# Line PLot:

plt.plot(df["Salary"], color = "red", marker = "*", linestyle = ":", linewidth = "2")
plt.grid()
plt.show()
```

# **Histogram**

A histogram is a type of plot that shows the distribution of a single (univariate) numerical variable by grouping values into ranges called bins and showing how many data points fall into each range.

## **Key idea**

**Bins:** Intervals (ranges) of your data

**Height of each bar:** Number of data points that fall in that bin (frequency)

**X-axis:** The value ranges (bins)

**Y-axis:** Frequency (count)

1.Splits it into bins (ranges)

2.Counts how many data points fall into each bin

3.Draws bars for those counts

**Why use a histogram**

To understand the shape of your data (normal, skewed, uniform, etc.)

To find outliers

To choose appropriate statistical models

```python
# Histogram

plt.hist(df["Salary"], bins = 5, color = "green")
plt.show()
```

# **Box Plot**

Key parts of a box plot

A standard box plot shows:

**Minimum →** Smallest value (excluding outliers)

**Q1 (First Quartile) →** 25% of data is below this value

**Median (Q2) →** Middle value (50% of data is below and above this)

**Q3 (Third Quartile) →** 75% of data is below this value

**Maximum →** Largest value (excluding outliers)

Additionally:

**Whiskers →** Extend from the box to the min and max values (excluding outliers)

**Outliers →** Data points far from the rest, shown as dots or stars

**Why use a box plot :**

Shows spread (variability) of data

Identifies outliers quickly

Compares distributions between multiple datasets

Works well for skewed data

```python
# Boxplot:

plt.boxplot(df["Salary"])
plt.show()
```

# **Pie Chart**

A pie chart is a circular chart that **divides data into slices**, where each slice’s size represents its proportion or **percentage** of the whole.

**Key idea**

The whole circle = **100%** of the data

Each slice = a category’s contribution to the total

Angles & areas of **slices** are proportional to their share

**When to use a pie chart : **

To show percentage composition of categories

When you have few categories (ideally ≤ 5–6)

When you want a quick visual impression of proportions

**⚠ Avoid pie charts if you have too many categories or small differences — a bar chart is often clearer.**

```python
# Pie chart:

count = df["dept"].value_counts()
count
```

```python
plt.pie(count, labels = count.index, autopct= "%1.2f", explode = [0, 0.1, 0.2])
plt.axis("equal")
plt.show()
```

# **Bar Plot**

A bar plot (or bar chart) is a graph that represents **categorical data** using rectangular bars, where the length/height of each bar corresponds to the value it represents.

**Key idea**

**X-axis →** Categories (e.g., days, products, countries)

**Y-axis →** Numeric values (counts, averages, totals, etc.)

**Bars →** Height (vertical bar chart) or length (horizontal bar chart) shows the value

**Why use a bar plot**

Compare values across categories

Show ranking or order of categories

Display differences clearly at a glance

```python
# countplot

plt.bar(count.index, count)
plt.show()
```

# **Scatter plot**

A scatter plot is a chart that displays data points as dots on a two-dimensional plane, showing the **relationship between two numerical variables**.

Key idea

**X-axis →** One **numerical** variable

**Y-axis →** Another **numerical** variable

**Each dot →** One observation (pair of values)

Can reveal correlation, clusters, and outliers

**When to use a scatter plot**

To see if two variables have a relationship (positive, negative, none)

To detect patterns, clusters, or outliers

To explore data before building models

```python
# Scatter Plot:

plt.scatter(df["Age"], df["Salary"], color = "orange", s=70, edgecolor="black")
plt.show()
```

# **4. Bivariate: Numerical - categorical**

```python
hr_sal = df[df["dept"]=="HR"]["Salary"]
it_sal = df[df["dept"] == "IT"]["Salary"]
fin_sal = df[df["dept"] == "Finance"]["Salary"]
```

# Box Plot

```python
plt.boxplot([hr_sal,it_sal,fin_sal],labels=["HR","IT","Finance"])
plt.show()
```

# Pie Chart

```python
# pie chart:
salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept,labels=salary_by_dept.index,autopct="%1.2f %%",explode=[0.1,0,0.1],shadow = True)
plt.show()
```

# Bar Plot

```python
# bar plot

hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
fin_mean = sum(fin_sal)/len(fin_sal)
```

```python
plt.bar(["HR", "IT", "Finance"], [hr_mean, it_mean, fin_mean], color = ["green", "black", "red"])
plt.grid()
plt.show()
```

# Bubble Plot(Scatter Plot)

```python
# Bubble PLot: Nothing but Scatter Plot

plt.scatter(df["Age"], df["Salary"], s = df["experience"]*50, color = "skyblue", edgecolors= "black")
plt.title("Age vs Salary vs Experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

# **6. 2 numerical and 1 categorical column:**

```python
plt.scatter(df["Age"], df["Salary"], c = df["dept"].map({"HR": "yellow", "IT":"blue", "Finance": "orange"}))
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary vs Dept")
plt.show()
```

```python
color = {"HR": "yellow", "IT":"blue", "Finance": "orange"}

for dept, color in color.items():
    df_dept = df[df["dept"] == dept]
    plt.scatter(df_dept["Age"], df_dept["Salary"], c = color, label = dept)

plt.legend()
```

# **7. Object Oriented API**

```python
fig, axs = plt.subplots(1,3,figsize=(15,5))

axs[0].plot(df["Age"], df["Salary"], color = "red", marker = "*", linewidth = "2")
axs[1].hist(df["Salary"],bins=5,color="skyblue")
axs[2].boxplot(df["Salary"])

plt.show()
```

# Load Fifa Data

```python
fifa = pd.read_csv('fifa_data.csv')

fifa.head(5)
```

# Histograms

```python
bins = [40,50,60,70,80,90,100]

plt.figure(figsize=(8,5))

plt.hist(fifa.Overall, bins=bins, color='#abcdef')

plt.xticks(bins)

plt.ylabel('Number of Players')
plt.xlabel('Skill Level')
plt.title('Distribution of Player Skills in FIFA 2018')

plt.show()
```

# Pie Chart

```python
left = fifa.loc[fifa['Preferred Foot'] == 'Left'].count()[0]
right = fifa.loc[fifa['Preferred Foot'] == 'Right'].count()[0]

plt.figure(figsize=(8,5))

labels = ['Left', 'Right']
colors = ['#abcdef', '#aabbcc']

plt.pie([left, right], labels = labels, colors=colors, autopct='%.2f %%')

plt.title('Foot Preference of FIFA Players')

plt.show()
```
